```python
# EDA - Outlier Analysis using IQR
outlier_summary = []
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    outliers = ((df[col] < lower_limit) | (df[col] > upper_limit)).sum()
    outlier_summary.append([
        col,
        Q1,
        Q3,
        lower_limit,
        upper_limit,
        outliers
    ])
outlier_summary = pd.DataFrame(
    outlier_summary,
    columns=[
        'Variable',
        'Q1',
        'Q3',
        'Lower_Limit',
        'Upper_Limit',
        'Outlier_Count'
    ]
)
outlier_summary
```

# Correlation Insights
- Total_Sales has a very strong positive correlation with Revenue (0.993), indicating that the two variables move almost together.
- Unit_Price has a moderate-to-strong positive correlation with Revenue (0.656).
- Units_Sold also has a moderate-to-strong positive correlation with Revenue (0.650).
- Store_Rating has almost no linear correlation with Revenue (0.003).
- Stock_On_Hand has almost no linear correlation with Revenue (-0.001).
- Discount_Percentage has a weak negative correlation with Revenue (-0.090).
- The correlation analysis indicates that Unit_Price and Units_Sold have substantially stronger linear relationships with Revenue than Store_Rating, Stock_On_Hand, or Discount_Percentage.
- Correlation represents linear association and does not establish causation.
- Total_Sales requires special attention during predictive modelling because of its extremely high correlation with Revenue. Its use as a predictor may cause target leakage if Total_Sales is calculated from information that directly determines Revenue.

```python
# Train-Test Split for Time Series Forecasting

train_size = int(len(monthly_revenue)*0.80)
train = monthly_revenue.iloc[:train_size]
test = monthly_revenue.iloc[train_size:]

print("Training Period: ",train.index.min(),"to",train.index.max())
print("Testing Period: ",test.index.min(),"to",test.index.max())

print("\nTraining Observations: ",len(train))
print("Testing Observations: ",len(test))
```

->Training period: Jan 2023 → Jul 2024
->Training observations: 19 months
->Testing period: Aug 2024 → Dec 2024
->Testing observations: 5 months
->The split is time-based, which is correct for forecasting because we should not randomly shuffle time-series data.
->We will train the forecasting model only on the first 19 months and evaluate it on the final 5 months.

```python
# Train the Forecasting Model
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
import numpy as np
import matplotlib.pyplot as plt

# Create Time Index
train_X = np.arange(len(train)).reshape(-1,1)
test_X = np.arange(len(train),len(monthly_revenue)).reshape(-1,1)

# Target Values (train and test are already Series containing only the Revenue values)
train_Y = train.values
test_Y = test.values

# Create and Train Model
model = LinearRegression()
model.fit(train_X,train_Y)

# Predict test period
test_predictions = model.predict(test_X)

# Display predictions
forecast_results = pd.DataFrame({
    'Actual_Revenue':test_Y,
    'Predicted_Revenue':test_predictions
}, index=test.index)

forecast_results
```

->August 2024: Actual ₹1.795B vs predicted ₹1.694B → model underestimated.
->September: Actual and predicted are almost identical.
->October: Actual ₹1.776B vs predicted ₹1.691B → significant underestimation.
->November: Prediction is very close to actual.
->December: Model again underestimates the actual revenue.
->The predictions are staying around ₹1.69B, while actual revenue fluctuates more.
->This tells us that Linear Regression captures the overall trend but does not capture monthly fluctuations/seasonality very well.

```python
# Forecasting accuracy metrics
mae = mean_absolute_error(test_Y,test_predictions)
rmse = np.sqrt(mean_squared_error(test_Y,test_predictions))
r2 = r2_score(test_Y, test_predictions)

print("Mean Absolute Error (MAE): ",mae)
print("Root Mean Squared Error (RMSE): ",rmse)
print("R-Squared (R^2):",r2)
```

->MAE: ₹49.22 million ≈ ₹4.92 crore
->RMSE: ₹63.51 million ≈ ₹6.35 crore
->R^2: -1.0606
->The negative R² means the Linear Regression model is not explaining the monthly revenue variation well on the test period.
->The model predicts values close to ₹1.69B every month, while actual revenue moves considerably up and down.
->From the previous graph, this dataset clearly has monthly fluctuations, which simple Linear Regression cannot capture.

```python
# Improved Forecasting Model
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
import numpy as np
import pandas as pd

# Train Holt's Exponential Smoothing model directly on the Series
holt_model = ExponentialSmoothing(
    train,
    trend='add',
    seasonal=None
).fit()

# Forecast the test period
holt_predictions = holt_model.forecast(len(test))

# Store results
holt_forecast_results = pd.DataFrame({
    'Actual_Revenue': test.values,
    'Predicted_Revenue': holt_predictions.values
}, index=test.index)

holt_forecast_results
```

```python
# Evaluate Holt's forecasting model
holt_mae = mean_absolute_error(
    test,
    holt_predictions
)

holt_rmse = np.sqrt(
    mean_squared_error(
        test,
        holt_predictions
    )
)

holt_r2 = r2_score(
    test,
    holt_predictions
)

print("Holt's Model MAE: ", holt_mae)
print("Holt's Model RMSE: ", holt_rmse)
print("Holt's Model R^2: ", holt_r2)
```

```python
# Compare forecasting models

comparison = pd.DataFrame({
    'Model': ['Linear Regression', "Holt's Exponential Smoothing"],
    'MAE': [mae, holt_mae],
    'RMSE': [rmse, holt_rmse],
    'R^2': [r2, holt_r2]
})

comparison
```

```python
# Model performance Comparison

plt.figure(figsize=(10,6))

plt.bar(comparison['Model'],comparison['MAE'])

plt.title('Forecasting Model Comparison: MAE')
plt.xlabel('Model')
plt.ylabel('Mean Absolute Error')

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Select the best forecasting model based on MAE

best_model_name = comparison.loc[
    comparison['MAE'].idxmin(), 'Model'
]

print("Best Forecasting Model:", best_model_name)
```

```python
# Train the selected model using all available data
full_revenue = monthly_revenue

if best_model_name == "Linear Regression":

    full_X = np.arange(len(full_revenue)).reshape(-1, 1)

    final_model = LinearRegression()
    final_model.fit(full_X, full_revenue.values)

    future_X = np.arange(
        len(full_revenue),
        len(full_revenue) + 6
    ).reshape(-1, 1)

    future_predictions = final_model.predict(future_X)

else:

    final_model = ExponentialSmoothing(
        full_revenue,
        trend='add',
        seasonal=None
    ).fit()

    future_predictions = final_model.forecast(6).values


# Create future dates
future_dates = pd.date_range(
    start=monthly_revenue.index[-1] + pd.DateOffset(months=1),
    periods=6,
    freq='ME'
)

# Create final forecast table
future_forecast = pd.DataFrame({
    'Date': future_dates,
    'Forecasted_Revenue': future_predictions
})

future_forecast
```
